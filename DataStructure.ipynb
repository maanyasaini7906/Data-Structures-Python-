```python
'''
1) What are data structures, and why are they important?
Ans. Data structure is a way of organising and storing data.
     They are important because the data can be accessed and manipulated easily.
2.) Explain the difference between mutable and immutable data types with
    examples.
Ans. Mutable data types can be changed after creation and immutabute data types
     cannot be changed.Example:mutable-lists,immutable-tuple
3.) Describe how dictionaries store data.
Ans. They store data in key-value pairs.
     dict={key:value}
4.) Why might you use a set instead of a list in Python?
Ans. Because it provides unique elements and faster membership tests.
5.) What is a string in Python, and how is it different from a list?
Ans. String is immutable and list is mutable.String is simply a sequence of
     characters.
6.) How do tuples ensure data integrity in Python?
Ans. By being immutable.
7.) What are the main differences between lists and tuples in Python?
Ans. Lists are mutable and '[]' are used to store elements.
     Tuples are immutable and '()' are used to store elements.
8.) What is a hash table, and how does it relate to dictionaries in Python?
Ans.A hash table is a data structure that stores key-value pairs using hash
    functions to quickly find keys. In python,dictionary is implemented using
    hash table.
9.) Can lists contain different data types in Python?
Ans. Yes.
10.) Explain why strings are immutable in Python?
Ans. Because the string cannot be modified.
11.) What advantages do dictionaries offer over lists for certain tasks?
Ans. In dictionary there is key-value concept and no need to search values
     using index,it can directly be done by keys.
12.) Describe a scenario where using a tuple would be preferable over a list.
Ans. Use a tuple instead of a list when you need to store fixed, unchangeable
     data.
     Scenario:
     Storing geographic coordinates (latitude, longitude) where the values
     should not change.
13.) How do sets handle duplicate values in Python?
Ans. Sets automatically removes duplicated values.
14.) How does the “in” keyword work differently for lists and dictionaries?
Ans. In lists: Checks if a value exists.
     In dictionaries: Checks if a key exists (not a value).
15.) Can you modify the elements of a tuple? Explain why or why not?
Ans. No, because tuples are immutable.
16.) What is a nested dictionary, and give an example of its use case.
Ans. A nested dictionary is a dictionary within a dictionary.
     students = {
         "101": {"name": "Alice", "age": 20},
         "102": {"name": "Bob", "age": 22}
           }
      Here, each student ID maps to another dictionary with their info.
17.)  Describe the time complexity of accessing elements in a dictionary.
Ans. Time complexity - O(1)
18.) In what situations are lists preferred over dictionaries?
Ans. Lists are preferred over dictionaries when:You're working with sequences
     of values without unique keys and you need indexed access.
19.) Why are dictionaries considered unordered, and how does that affect
     data retrieval?
Ans. Dictionaries in Python were historically unordered because they stored
     items without maintaining the order of insertion.
     Iterating over a dictionary would return keys in an unpredictable order.
20.) Explain the difference between a list and a dictionary in terms of
     data retrieval.
Ans. List:
     Retrieve data by index (position).
     Dictionary:
     Retrieve data by key (unique identifier).
'''
```
